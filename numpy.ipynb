```python

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Find the average age of students.
age = data[0:,0]
print(age)
```

```python
avg = age.mean()
print(avg)
```

```python
#Extract Math marks of all students.
math = data[0:,1]
print(math)
```

```python
#Find the highest Science mark.
Science = data[0:,2]
print(Science)
```

```python
highest_science =Science.max()
print(highest_science)
```

```python
#Get details of the student who scored more than 90 in Math.
for i in range(len(data)):
     if math[i] >= 90:
        print(data[i])
```

```python
#Increase Math marks of all students by 5.
math = data[0:,1]
print(math)
```

```python
math+=5
print(math)
```
